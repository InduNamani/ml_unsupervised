# Hierarchical Clustering - Iris Dataset

```python
# importing libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
```

```python
# loading the dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
# basic info
print(df.shape)
print(df.isnull().sum())
```

```python
# scaling the data
scaler = StandardScaler()
X = scaler.fit_transform(df)
print("scaling done")
```

```python
# dendrogram helps us decide how many clusters to pick
linked = linkage(X, method='ward')

plt.figure(figsize=(10, 5))
dendrogram(linked, truncate_mode='lastp', p=20)
plt.title('Dendrogram')
plt.xlabel('Samples')
plt.ylabel('Distance')
plt.show()
# look for the longest vertical line that doesnt get cut - thats where we cut
```

```python
# from dendrogram K=3 looks right, applying agglomerative clustering
model = AgglomerativeClustering(n_clusters=3, linkage='ward')
labels = model.fit_predict(X)
df['Cluster'] = labels
df.head()
```

```python
# cluster counts
print(df['Cluster'].value_counts())
```

```python
# silhouette score
score = silhouette_score(X, labels)
print("Silhouette Score:", round(score, 4))
```

```python
# plotting the clusters
colors = ['red', 'blue', 'green']
for i in range(3):
    points = X[labels == i]
    plt.scatter(points[:, 0], points[:, 1], c=colors[i], label=f'Cluster {i}')

plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title('Hierarchical Clustering')
plt.legend()
plt.show()
```

```python
# mean values per cluster
df.groupby('Cluster').mean().round(2)
```

### Conclusion

- hierarchical clustering doesnt need K upfront, dendrogram helps us decide
- used ward linkage which minimizes variance within clusters
- got 3 clusters similar to actual iris species
- silhouette score confirms clusters are well separated
- unlike kmeans, no random initialization so results are consistent
