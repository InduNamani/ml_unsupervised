# t-SNE on Digits Dataset
Simple implementation of t-SNE for visualization.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.manifold import TSNE

df = pd.read_csv('../data/digits.csv')
X = df.drop('target', axis=1).values
y = df['target'].values
print('Shape:', df.shape)
```

```python
X_scaled = StandardScaler().fit_transform(X)
# Pre-reduce with PCA for speed
X_pre = PCA(n_components=50, random_state=42).fit_transform(X_scaled)

tsne = TSNE(n_components=2, perplexity=30, n_iter=1000,
            learning_rate=200, init='pca', random_state=42)
X_tsne = tsne.fit_transform(X_pre)
print('KL divergence:', round(tsne.kl_divergence_, 4))
```

```python
plt.figure(figsize=(9,7))
for d in range(10):
    mask = y==d
    plt.scatter(X_tsne[mask,0], X_tsne[mask,1], label=str(d), alpha=0.8, s=18)
plt.xlabel('t-SNE 1'); plt.ylabel('t-SNE 2')
plt.title(f't-SNE - Digits Dataset (KL={tsne.kl_divergence_:.3f})')
plt.legend(title='Digit', ncol=2, fontsize=8)
plt.tight_layout()
plt.show()
print('Done!')
```
