# PCA on Digits Dataset
Simple implementation of PCA for dimensionality reduction.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

df = pd.read_csv('../data/digits.csv')
X = df.drop('target', axis=1).values
y = df['target'].values
print('Shape:', df.shape)
```

```python
X_scaled = StandardScaler().fit_transform(X)

pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_scaled)
print('Variance explained:', pca.explained_variance_ratio_.sum().round(3))
```

```python
plt.figure(figsize=(8,6))
for d in range(10):
    mask = y==d
    plt.scatter(X_pca[mask,0], X_pca[mask,1], label=str(d), alpha=0.7, s=15)
plt.xlabel('PC1'); plt.ylabel('PC2')
plt.title('PCA - Digits Dataset')
plt.legend(title='Digit', ncol=2, fontsize=8)
plt.tight_layout()
plt.show()
```

```python
# Scree plot
pca_full = PCA(random_state=42).fit(X_scaled)
cumvar = np.cumsum(pca_full.explained_variance_ratio_)
plt.figure(figsize=(8,4))
plt.plot(range(1, len(cumvar)+1), cumvar*100, 'o-', color='steelblue')
plt.axhline(95, color='red', linestyle='--', label='95%')
plt.xlabel('Components'); plt.ylabel('Cumulative Variance (%)')
plt.title('Cumulative Explained Variance')
plt.legend(); plt.tight_layout(); plt.show()
print('Done!')
```
