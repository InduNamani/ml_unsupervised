# 📊 Hierarchical Clustering - Customer Segmentation

This notebook uses a customer dataset with Income, Spending Score, and Age.


```python
import pandas as pd
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import dendrogram, linkage
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
df = pd.read_csv('../data/customers.csv')
df.head()
```

```python
X = df[['Annual_Income','Spending_Score','Age']]
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
linked = linkage(X_scaled, method='ward')
plt.figure(figsize=(10,5))
dendrogram(linked)
plt.show()
```

```python
model = AgglomerativeClustering(n_clusters=3)
labels = model.fit_predict(X_scaled)
df['Cluster'] = labels
df.head()
```

```python
plt.scatter(X['Annual_Income'], X['Spending_Score'], c=labels)
plt.xlabel('Income')
plt.ylabel('Spending Score')
plt.show()
```

```python
print('Silhouette Score:', silhouette_score(X_scaled, labels))
```
