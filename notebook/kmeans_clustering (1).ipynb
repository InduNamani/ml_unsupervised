# K-Means Clustering - Iris Dataset

```python
# importing the libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
# loading the dataset
iris = load_iris()
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df.head()
```

```python
# basic info
print(df.shape)
print(df.describe())
```

```python
# checking missing values
df.isnull().sum()
```

```python
# scaling the data before applying kmeans
scaler = StandardScaler()
X = scaler.fit_transform(df)
print("done")
```

```python
# finding the right K using elbow method
wcss = []
for k in range(1, 11):
    km = KMeans(n_clusters=k, random_state=42)
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1, 11), wcss, 'bo-')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')
plt.title('Elbow Method')
plt.show()
```

```python
# from the elbow plot K=3 looks good, so lets go with that
kmeans = KMeans(n_clusters=3, random_state=42)
kmeans.fit(X)
df['Cluster'] = kmeans.labels_
df.head()
```

```python
# how many points in each cluster
print(df['Cluster'].value_counts())
```

```python
# silhouette score to check how good the clusters are
score = silhouette_score(X, kmeans.labels_)
print("Silhouette Score:", round(score, 4))
# score closer to 1 means clusters are well separated
```

```python
# visualizing the clusters using first 2 features
colors = ['red', 'blue', 'green']
for i in range(3):
    points = X[kmeans.labels_ == i]
    plt.scatter(points[:, 0], points[:, 1], c=colors[i], label=f'Cluster {i}')

# plotting centroids too
centers = kmeans.cluster_centers_
plt.scatter(centers[:, 0], centers[:, 1], c='black', marker='*', s=200, label='Centroids')

plt.xlabel(iris.feature_names[0])
plt.ylabel(iris.feature_names[1])
plt.title('K-Means Clusters')
plt.legend()
plt.show()
```

```python
# mean values of each cluster
df.groupby('Cluster').mean().round(2)
```

### Conclusion

- used iris dataset with 150 samples and 4 features
- applied standard scaling before kmeans (important step)
- elbow method helped us pick K=3
- silhouette score shows clusters are reasonably well separated
- cluster 0, 1, 2 roughly match the 3 iris species
